# Pruebas del simulador CPU
Pruebas exploratorias con un conjunto fijo de procesos. Ejecuta desde la carpeta raíz del proyecto.

```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from algoritmos.fcfs import fcfs
from algoritmos.sjf import sjf
from algoritmos.round_robin import round_robin
from algoritmos.prioridades import prioridades
from simulacion.metricas import construir_resumen

procesos = [
    {'id': 'P1', 'llegada': 0, 'rafaga': 8, 'prioridad': 2},
    {'id': 'P2', 'llegada': 1, 'rafaga': 4, 'prioridad': 1},
    {'id': 'P3', 'llegada': 2, 'rafaga': 6, 'prioridad': 3},
    {'id': 'P4', 'llegada': 3, 'rafaga': 2, 'prioridad': 2},
]
resultados = {
    'FCFS': fcfs(procesos),
    'SJF': sjf(procesos),
    'Round Robin': round_robin(procesos, quantum=2),
    'Prioridades': prioridades(procesos),
}
construir_resumen(resultados).round(2)
```

```python
for nombre, resultado in resultados.items():
    print(f'\n{nombre}')
    print('Gantt:', resultado['gantt'])
    print('Detalle:', resultado['detalle'])
```
